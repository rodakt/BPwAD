## 2. Gradient descent w 2D

Przechodzimy do funkcji dwóch zmiennych. Gradient jest teraz wektorem dwóch pochodnych cząstkowych:

$$\nabla f(x, y) = \left[\frac{\partial f}{\partial x},\; \frac{\partial f}{\partial y}\right]$$

a reguła aktualizacji to:

$$\mathbf{w}^{(\tau)} = \mathbf{w}^{(\tau-1)} - \eta \, \nabla f(\mathbf{w}^{(\tau-1)})$$

gdzie $\mathbf{w} = [x, y]$.

### 2.1 Funkcja Rosenbrocka

Funkcja Rosenbrocka to klasyczny benchmark optymalizacji:

$$f(x, y) = (1 - x)^2 + 100(y - x^2)^2$$

Ma jedno minimum globalne w $(1, 1)$, ale otoczone jest wąską, zakrzywioną doliną, która sprawia kłopoty algorytmom optymalizacji.

**Dlaczego Rosenbrock jest trudny?** Dotarcie do punktu $(1, 1)$ wymaga przejścia dnem zakrzywionej doliny. Gradient wskazuje w stronę dna doliny (prostopadle do jej osi), ale niekoniecznie wzdłuż niej - algorytm oscyluje między ścianami zamiast posuwać się naprzód. To jeden z motywów, dla których wymyślono momentum (sekcja 4).

## 4. Momentum

Algorytm gradientu prostego ma tendencję do oscylacji w wąskich dolinach (jak na Rosenbrocku) - gradient jest duży prostopadle do dna doliny, ale mały wzdłuż niej. **Momentum** łagodzi ten problem, dodając „bezwładność": aktualizacja uwzględnia nie tylko bieżący gradient, ale też kierunek z poprzednich kroków.

### 4.1 Algorytm

Gradient descent z momentum utrzymuje wektor prędkości $\mathbf{v}$:

$$\mathbf{v}^{(\tau)} = \beta\,\mathbf{v}^{(\tau-1)} + \nabla f(\mathbf{w}^{(\tau-1)})$$
$$\mathbf{w}^{(\tau)} = \mathbf{w}^{(\tau-1)} - \eta\,\mathbf{v}^{(\tau)}$$

gdzie $\beta \in [0, 1)$ to hiperparametr momentum. Inicjalizacja: $\mathbf{v}^{(0)} = \mathbf{0}$.

**Analogia fizyczna.** Wyobraź sobie kulkę toczącą się po powierzchni funkcji. Bez momentum kulka natychmiast reaguje na nachylenie terenu - na grani doliny zmienia kierunek co krok. Z momentum kulka ma masę: gromadzi prędkość w kierunku, w którym gradient jest spójny (wzdłuż doliny), a oscylacje prostopadłe wzajemnie się znoszą, bo gradient zmienia tam znak co krok. Momentum pozwala kulce „przeskoczyć" przez wąskie grzbiety i szybciej dotrzeć do minimum.
